```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)
```

```python
data_path = Path("../data/bronze")
```

```python
customers = pd.read_csv(data_path / "customer.csv")
products = pd.read_csv(data_path / "product.csv", on_bad_lines="skip")
transactions = pd.read_csv(data_path / "transactions.csv")
clickstream = pd.read_csv(data_path / "click_stream.csv")
```

```python
print(f"\n Customers: {len(customers):,} rows")
print(f" Products: {len(products):,} rows")
print(f" Transactions: {len(transactions):,} rows")
print(f" Clickstream: {len(clickstream):,} rows")
```
